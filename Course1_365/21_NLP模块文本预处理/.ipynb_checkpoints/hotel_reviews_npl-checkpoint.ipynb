# 酒店评论文本预处理：小写转换

## Goal

读取 `tripadvisor_hotel_reviews.csv`，将 `Review` 列统一转换为小写。

## Setup

导入 pandas，并设置数据文件路径。原始 CSV 不会被修改。

```python
from pathlib import Path

import pandas as pd


DATA_PATH = Path("tripadvisor_hotel_reviews.csv")
DATA_PATH
```

## Steps

### 1. 读取评论数据

```python
hotel_reviews = pd.read_csv(DATA_PATH)

print(f"数据行数：{len(hotel_reviews)}")
print(f"列名：{hotel_reviews.columns.tolist()}")
hotel_reviews.head(3)
```

### 2. 将评论转换为小写

使用 pandas 的字符串方法 `.str.lower()`，统一英文大小写，方便后续分词和统计。

```python
# 保留一份处理前的评论，便于对比。
review_before = hotel_reviews["Review"].copy()

# 将 Review 列中的英文字符统一转换为小写。
hotel_reviews["Review"] = hotel_reviews["Review"].astype("string").str.lower()
```

### 3. 查看处理前后对比

```python
comparison = pd.DataFrame({
    "处理前": review_before.str.slice(0, 120),
    "处理后": hotel_reviews["Review"].str.slice(0, 120),
})

comparison.head(5)
```

## Checks

确认行数没有变化，并检查处理后的评论是否还包含大写字母。

```python
contains_uppercase = hotel_reviews["Review"].str.contains(r"[A-Z]", regex=True, na=False).any()

assert len(hotel_reviews) == len(review_before), "处理前后行数不一致"
assert not contains_uppercase, "处理后的评论中仍包含大写字母"

print(f"✅ 小写处理完成：{len(hotel_reviews)} 条评论")
print("✅ 检查通过：处理后未发现大写英文字母")
```
