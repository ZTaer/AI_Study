# NLP 文本预处理入门

## Goal

把一段原始英文评论转换成更适合机器处理的词语列表。

本练习依次完成：**小写化 → 去除标点 → 分词 → 去除停用词**。为了便于理解，代码只使用 Python 标准库。

## Setup

先准备一条酒店评论。预处理前，文本中包含大小写、标点和一些常见词。

```python
text = "The room was not large, but it was clean and comfortable! Parking was expensive."
print("原始文本：", text)
```

## Steps

### 1. 转成小写

`The` 和 `the` 对人来说通常是同一个词，但计算机默认会把它们当成不同字符串。统一为小写可以减少这种重复。

```python
lower_text = text.lower()
print("小写化：", lower_text)
```

### 2. 去除标点

这里用正则表达式删除逗号、感叹号和句号，只保留字母、数字、下划线和空白字符。

```python
import re

clean_text = re.sub(r"[^\w\s]", "", lower_text)
print("去除标点：", clean_text)
```

### 3. 分词

分词（Tokenization）就是把一整段文本拆成单独的词。这个英文示例用 `split()` 按空白切分即可。

```python
tokens = clean_text.split()
print("分词结果：", tokens)
```

### 4. 去除停用词

停用词是出现频率很高、但在某些任务中信息量较少的词，例如 `the`、`was`。

注意：这里**故意保留 `not`**。在情感分析中，`not good` 和 `good` 意思相反，盲目删除 `not` 会丢失重要语义。

```python
# 这里只列出示例文本中需要删除的几个常见词；not 不在集合中。
stop_words = {"the", "was", "but", "it", "and"}
filtered_tokens = [word for word in tokens if word not in stop_words]

print("去除停用词：", filtered_tokens)
print("最终文本：", " ".join(filtered_tokens))
```

## Checks

把刚才的步骤封装成函数，再用断言检查两个关键结果：标点已删除，并且 `not` 被保留。

```python
def preprocess_text(raw_text):
    lower_text = raw_text.lower()
    clean_text = re.sub(r"[^\w\s]", "", lower_text)
    tokens = clean_text.split()
    return [word for word in tokens if word not in stop_words]

result = preprocess_text(text)

expected_result = ["room", "not", "large", "clean", "comfortable", "parking", "expensive"]
assert result == expected_result, "预处理结果与预期不一致"
print("检查通过：", result)
```
